```python
from typing import Literal
from pydantic import BaseModel, ConfigDict, Field, field_validator

class Salida(BaseModel):
    model_config = ConfigDict(extra="forbid")

    id_paquete: str = Field(min_length=1)
    categoria: str = Literal["normal", "urgente"]
    confianza: float = Field(ge=0, le=1)
```

```python
salidas = [Salida(id_paquete="P02", categoria="normal", confianza=0.5)]

salida = Salida(id_paquete="P02", categoria="normal", confianza=0.5)
salida2 = Salida(id_paquete="P03", categoria="normal", confianza=0.6)

df = pd.DataFrame(columns=["id_paquete", "categoria", "confianza"])

entry = pd.DataFrame.from_dict({
     "id_paquete": [salida.id_paquete],
     "categoria":  [salida.categoria],
     "confianza": [salida.confianza]
})

df = pd.concat([df, entry], ignore_index=True)

entry = pd.DataFrame.from_dict({
     "id_paquete": [salida2.id_paquete],
     "categoria":  [salida2.categoria],
     "confianza": [salida2.confianza]
})
df = pd.concat([df, entry], ignore_index=True)

df.head()
```
